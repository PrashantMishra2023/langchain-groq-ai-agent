```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature = 0.6)

prompt_template_name = PromptTemplate(
    input_variables=["cuisine"],
    template = "I want to open a restaurant for {cuisine} food. Can you suggest me some good names for it?"
)

chain = prompt_template_name | llm

prompt_template_items = PromptTemplate(
    input_variables=["restaurent_name"],
    template = "Suggest some menu items for {restaurent_name}. Return it as a comma separated."
)

food_items_chain = prompt_template_items | llm

```

```python
name_response = chain.invoke({"cuisine": "Indian"})
restaurent_name = name_response.content

print("Restaurent Name: ", restaurent_name)
```

```python
food_response = food_items_chain.invoke({"restaurent_name": restaurent_name})

print("\n Menu Items: ", food_response.content)
```

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
name_prompt = ChatPromptTemplate.from_template(
    "Suggest a fancy restaurant name for {cuisine} food."
)

menu_prompt = ChatPromptTemplate.from_template(
    "Suggest 5 menu items for {restaurant_name}."
)

name_chain = name_prompt | llm
menu_chain = menu_prompt | llm
```

```python
from langchain_core.runnables import RunnableLambda

simple_chain = (
    name_chain
    | RunnableLambda(lambda x: {"restaurant_name": x.content})
    | menu_chain
)

result = simple_chain.invoke({"cuisine": "Mexican"})

print(result.content)
```

```python
from langchain_core.runnables import RunnableParallel

sequential_chain = (
    name_chain
    | RunnableLambda(lambda x: {"restaurant_name": x.content})
    | RunnableParallel(
        restaurant_name=lambda x: x["restaurant_name"],
        menu_items=menu_chain
    )
)

result = sequential_chain.invoke({"cuisine": "Indian"})

print("Restaurant Name:", result["restaurant_name"])
```

```python
print("Menu Items:", result["menu_items"].content)
```

```python
from secret import groq_api_key
import os
os.environ["GROQ_API_KEY"] = groq_api_key

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

print("Groq LLM ready!")
```

```python
from langchain.tools import tool

@tool
def calculator(expression: str) -> str:
    """Calculate a mathematical expression."""
    
    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return str(result)
    except Exception as e:
        return f"Error: {e}"
```

```python
from langchain.agents import create_agent

tools = [calculator]

agent = create_agent(
    model=llm,
    tools=tools
)

print("Agent created successfully!")
```

```python
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "Calculate 25 * 10 + 50"
        }
    ]
})

print(response["messages"][-1].content)
```
